## Study

### efetch
[Documentation on Entrez.efetch parameters](https://www.ncbi.nlm.nih.gov/books/NBK25499/#chapter4.EFetch)

Table of possible return type `rettype` arguments: [here](www.ncbi.nlm.nih.gov/books/NBK25499/table/chapter4.T._valid_values_of__retmode_and/?report=objectonly)

```python
Bio.Entrez.email="darryl.fung.professional@gmail.com"
handle=Bio.Entrez.efetch(db='nucleotide',
                        id=['FJ817486, JX469983'],
                        rettype='fasta')
records=handle.read()
print(records)
```

For some information about Entrez.read() for XML files: https://biopython.org/docs/dev/Tutorial/chapter_entrez.html

We can use Python's normal read() for simpler text files like FASTA.

### SeqIO.parse

```python
help(Bio.SeqIO.parse)
```

```python
Bio.Entrez.email="darryl.fung.professional@gmail.com"
handle=Bio.Entrez.efetch(db='nucleotide',
                        id=['FJ817486, JX469983'],
                        rettype='fasta')
records=list(Bio.SeqIO.parse(handle,"fasta"))
    # list of SeqIO objects in FASTA format
print (records[0].id) #first record's ID (remember py starts at 0)

print (len(records[-1].seq))
    # length of the last record
    # NOTE that .seq is there because we invoke the sequence attribute of the list
```

```python
# len(seqs[0])
# seqs

seqs=[i.seq for i in records]
# seqs.index(min(seqs, key=len))
    # grabbing index number of the sequence with the shortest length.
    # Note that the indexes of seqs = indexes of records
```

```python
print(records[
    seqs.index(min(seqs, key=len))
])
# print the whole SeqIO object whose index matches the index of the shortest sequence length.
```

[Some more information about Bio.SeqRecord and the attributes that SeqRecord objects can hold](https://biopython.org/docs/1.75/api/Bio.SeqRecord.html)

## Execution

For n ID's, with n≤10, return the ID of the shortest sequence in __FASTA__ format.
- Note: FASTA format =/= human-readable format that we get from normal `print()`.
- (We use the format method of the SeqFeature object class. See below)

```python
Bio.Entrez.email="darryl.fung.professional@gmail.com"
handle=Bio.Entrez.efetch(db='nucleotide',
                        id=['NM_001159758,JX462670,JX398977,FJ817486,JX491654,NM_204821,JX462666,JQ011270,NM_001251956,JX317624'],
                        rettype='fasta')
records=list(Bio.SeqIO.parse(handle,"fasta"))

print(
    min(records,key=len)
.format("fasta"))
```

```python
print(
    min(records,key=len)
.format("fasta"))
```

Documentation about the `format` method of SeqRecord objects [here](https://biopython.org/wiki/SeqRecord).
- Here, we use this to print in the fasta format (instead of human-readable)
