Grab GenBank sequences as FASTA files from JX205496.1 and JX469991.1 (see `...fasta_efetch_seqIO.ipynb`)

### Entrez, SeqIO

```python
from Bio import Entrez, SeqIO
```

```python
Entrez.email = "darryl.fung.professional@gmail.com"

accessions = ["JX205496.1", "JX469991.1"]

with Entrez.efetch(
    db="nucleotide",
    id=accessions,
    rettype="fasta",
    retmode="text"
) as handle:

    records = list(SeqIO.parse(handle, "fasta"))

seqs=[i.seq for i in records]
    # print(record.id, record.seq)
```

### needle notes

needle requires .fasta input files.

To input into EMBOSS package's needle:
`needle -asequence sequence1_forneedle.fasta -bsequence sequence2_forneedle.fasta -gapopen 10 -gapextend 1 -endweight Y -endopen 10 -endextend 1 -outfile test_align_2`

*or just omit outfile and let the terminal prompt you for an output name*

**NOTE: needle has trouble overriding previously-written files. Make a new file name for each run**
