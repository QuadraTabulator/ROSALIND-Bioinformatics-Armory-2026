### Genus & Publication Date filtering
Genus: `term='"genus"[Organism]'`
- eg: all species under the *Drosophila* genus (*Drosophila appendiculata, Drosophila melanogaster*...)

Publication date: `datetype='pdat'` Use with min/maxdate (see below)

```python
Entrez.email="darryl.fung.professional@gmail.com"
handle=Entrez.esearch(db='nucleotide',
                      term='"Drosophila"[Organism]',
                      datetype='pdat',
                     mindate="2003/08/08",
                     maxdate="2010/08/08")
    # '"genus"[Organism]' for genus search
    # When using min/maxdate, must specify a datetype. eg: pdat for publication date
    # mindate & maxdate MUST be used together.
        # mindate and maxdate syntax: YYYY/MM/DD
record=Entrez.read(handle)
record["Count"]
```

### Note on Entrez vs SeqIO and Python parsing/reading methods
Important note: `Entrez.read` is used here because we are asking for outputs in structured XML format -- a format that we can obtain information like Count or IdList from. (Entrez.read points for Biopython to parse for us.)
- *If we were simply asking for raw text, we could instead use the Python `handle.read`. But for FASTA files as biological sequence objects, we should invoke the `SeqIO` module and use `SeqIO.parse` or 1SeqIO.read`*

## Execution

- Genus: Miconia

```python
Entrez.email="darryl.fung.professional@gmail.com"
handle=Entrez.esearch(db="nucleotide",
                      term='"Miconia"[Organism]',
                      datetype='pdat',
                      mindate="2005/12/23",
                      maxdate="2010/05/30")
record=Entrez.read(handle)
record['Count']
```
